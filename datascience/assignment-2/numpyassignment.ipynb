# NumPy Assignment

```python
import numpy as np
```

## 1. 1D array (1 to 20)

Basic stats first, then find where values are > 10.


```python
# numbers 1 through 20
arr = np.arange(1, 21)
print(arr)
```

```python
print("sum:", arr.sum())
print("mean:", arr.mean())
print("median:", np.median(arr))
print("std:", arr.std())
```

```python
# positions where value > 10 (not the values themselves)
idx = np.where(arr > 10)
print(idx)
```

## 2. 4x4 array

Make it, transpose it, then sum by rows and columns.


```python
arr2d = np.arange(1, 17).reshape(4, 4)
print(arr2d)
```

```python
# rows <-> columns
print(arr2d.T)
```

```python
# axis=1 → across each row, axis=0 → down each column
print("row sums:", np.sum(arr2d, axis=1))
print("col sums:", np.sum(arr2d, axis=0))
```

## 3. Two random 3x3 arrays

Element-wise ops, then a proper matrix multiply (dot).


```python
# random ints from 1 to 20
a = np.random.randint(1, 21, size=(3, 3))
b = np.random.randint(1, 21, size=(3, 3))
print("a:\n", a)
print("b:\n", b)
```

```python
print("add:\n", a + b)
print("sub:\n", a - b)
print("mul:\n", a * b) 
```

```python
print("dot product:\n", np.dot(a, b))
```

## 4. Reshape + slice

12 numbers → 3x4, then grab first 2 rows and last 2 cols.


```python
flat = np.arange(1, 13)
mat = flat.reshape(3, 4)
print(mat)
```

```python
# [:2] = first two rows, [-2:] = last two columns
print(mat[:2, -2:])
```
