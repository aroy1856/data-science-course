# Pandas Assignment

Working with a small employee dataframe — filter, group, sort, save to csv.


```python
import pandas as pd
```

## 1. Create the DataFrame

Then peek at the data, get Age/Salary stats, and avg salary for HR.


```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}
df = pd.DataFrame(data)
df
```

```python
# first 5 rows (we only have 5 anyway)
df.head()
```

```python
# statistics of Age and Salary
df[['Age', 'Salary']].describe()
```

```python
hr = df[df['Department'] == 'HR']
print("avg HR salary:", hr['Salary'].mean())
```

## 2. Bonus column

10% of salary.


```python
df['Bonus'] = df['Salary'] * 0.1
df
```

## 3. Filter by age

Only people between 25 and 30 (inclusive).


```python
df[(df['Age'] >= 25) & (df['Age'] <= 30)]
```

## 4. Average salary by department


```python
df.groupby('Department')['Salary'].mean()
```

## 5. Sort by salary + save CSV

Ascending order, then write to employee_data.csv.


```python
# ascending as per assignment
sorted_df = df.sort_values(by='Salary', ascending=True)
sorted_df.to_csv('employee_data.csv', index=False)
sorted_df
```
