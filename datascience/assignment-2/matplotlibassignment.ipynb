# Matplotlib Assignment

Just plotting some basic charts with matplotlib — line, bar, pie, and a histogram.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Line plot

Plotting x vs y, then adding title / labels / grid.

```python
# given data
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]
```

```python
plt.plot(x, y, label='values')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Plot')
plt.legend()
plt.grid(True)

plt.show()
```

## 2. Bar graph

Student marks — red bars with a title.

```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]
  
```

```python
plt.bar(students, marks, color='red')

plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Student Marks')

plt.show()
```

## 3. Pie chart

Revenue by region. North America is highest so I'm exploding that slice a bit.

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

# pull out the biggest slice (North America = index 0)
explode = [0.1, 0, 0, 0]

plt.figure(figsize=(6, 6))
plt.pie(revenue, labels=regions, autopct='%1.1f%%', explode=explode)
plt.title('Revenue by Region')

plt.show()
```

## 4. Histogram

1000 random ints from 1–100, then plot the frequency.

```python
data = np.random.randint(1, 101, size=1000)

plt.hist(data, bins=20, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Frequency Distribution (1 to 100)')
plt.show()
```
