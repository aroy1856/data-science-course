# Task 6: Filtering & Conditional Selection
Perform the following filters:
1. Students who scored more than 75 marks.
2. Students belonging to subject Math.
3. Students who scored more than average marks.
4. Students who failed (marks < 70).

````python
# students who scored more than 75 marks
students[students['Marks'] > 75]
````

````python
# students belonging to subject Math
students[students['Subject'] == 'Math']
````

````python
# students who scored more than average marks
students[students['Marks'] > students['Marks'].mean()]
````

````python
# students who failed (marks < 70)
students[students['Marks'] < 70]
````

# Task 7: Grouping & Basic Analyses

````python
# average marks per subject
students.groupby('Subject')['Marks'].mean()
````

````python
# count number of students per subject
students.groupby('Subject')['Name'].count()
````

````python
# maximum marks per subject
students.groupby('Subject')['Marks'].max()
````

# Task 9: Mini Use Case: Sales Data Analysis
Create a DataFrame:
```
sales = {
'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
'Revenue': [1200, 1500, 900, 2000, 1800]
}
```
Perform:
1. Total revenue
2. Average daily revenue
3. Day with highest revenue
4. Days where revenue > average
5. Plot revenue vs day

````python
sales = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
sales = pd.DataFrame(sales)
sales
````

````python
# total revenue
sales['Revenue'].sum()
````

````python
# avg daily revenue
sales['Revenue'].mean()
````

````python
# day with highest revenue
sales.loc[sales['Revenue'].idxmax()]['Day']
````

````python
# days where revenue > average
sales[sales['Revenue'] > sales['Revenue'].mean()]
````

````python
# plot revenue vs day
sales.plot(x='Day', y='Revenue')
````
