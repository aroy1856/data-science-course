# Task 5: Matrix Plots
1. Create a pair plot using sns.pairplot().
2. Create a heatmap of correlation matrix.

```python
# pair plot 
sns.pairplot(df)
plt.show()
```

```python
# heatmap of correlation matrix

numerical_cols = df.select_dtypes(include=['number']).columns
print(numerical_cols)
numerical_cols = numerical_cols.drop('Product_ID')
sns.heatmap(df[numerical_cols].corr(), annot=True)
plt.show()
```

# Task 8: Multi-Plots & Figure-Level Plots
1. Create a FacetGrid with:
   - One numerical variable on x-axis
   - One numerical variable on y-axis
   - Categorical column as col or row
2. Create a multi-plot dashboard using:
   - relplot
   - catplot
   - displot

```python
# Create a FacetGrid 
g = sns.FacetGrid(df, col="Product_Category")
g.map_dataframe(sns.scatterplot, x="Discount", y="Sales_Amount")
g.set_axis_labels("Discount", "Sales Amount")
plt.show()
```

```python
# relplot 
sns.relplot(data=df, x="Discount", y="Sales_Amount",hue="Product_Category",kind="scatter")
```

```python
# catplot 
sns.catplot(data=df, x="Product_Category", y="Sales_Amount",kind="box")
```

```python
# displot     
sns.displot(data=df,x="Sales_Amount", hue="Product_Category", kind="hist")
plt.show()
```
