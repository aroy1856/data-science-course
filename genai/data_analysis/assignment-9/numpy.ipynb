# Task 1: Creating NumPy Arrays
1. Import NumPy as np.
2. Create the following arrays:

   - A NumPy array from the list: [10, 20, 30, 40]
3. Print:

# Task 2: Important Mathematical Operations
Given two arrays:
   ```
   A = np.array([10, 20, 30, 40])
   B = np.array([1, 2, 3, 4])
   ```
Perform and print results for:
1. Addition (A + B)
2. Subtraction (A - B)
3. Multiplication (A * B)
4. Division (A / B)
5. Power (A ** 2)

- Extra (optional): Use NumPy's functions: np.add(), np.subtract() for the same operations.

````python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

print(f"A + B : {A+B}")
print(f"A - B : {A-B}")
print(f"A * B : {A*B}")
print(f"A / B : {A/B}")
print(f"A**2 : {A**2}")

print("--------------------------------")

print(f"A + B : {np.add(A,B)}")
print(f"A - B : {np.subtract(A,B)}")
print(f"A * B : {np.multiply(A,B)}")
print(f"A / B : {np.divide(A,B)}")
print(f"A**2 : {np.power(A,2)}")
````

# Task 3: Important NumPy Mathematical Formulas
Given:
```
values = np.array([2, 4, 6, 8, 10])
```
Calculate:
1. Square root of each element
2. Exponential of each element
3. Natural logarithm of each element
4. Sum of all elements
5. Cumulative sum of elements
- Use NumPy functions only.

````python
values = np.array([2, 4, 6, 8, 10])

print(f"Square root of each element: {np.sqrt(values)}")
print(f"Exponential of each element: {np.exp(values)}")
print(f"Natural logarithm of each element: {np.log(values)}")
print(f"Sum of all elements: {np.sum(values)}")
print(f"Cumulative sum of elements: {np.cumsum(values)}")
````

# Task 4: Aggregation Operations
Given a 2D array:
```
data = np.array([
[10, 20, 30],
[40, 50, 60],
[70, 80, 90]
])
```
Find:
1. Row-wise sum
2. Column-wise sum
3. Minimum value
4. Maximum value
5. Overall mean


````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("row wise sum: ", np.sum(data, axis=1))
print("column wise sum: ", np.sum(data, axis=0))
print("minimum value: ", np.min(data))
print("maximum value: ", np.max(data))
print("overall mean: ", np.mean(data))
````

# Task 5: Statistical Operations (Core Focus)
Given:
```
marks = np.array([78, 85, 90, 66, 72, 88, 95, 60])
```
Calculate and print:
1. Mean
2. Median
3. Variance
4. Standard Deviation
5. Minimum & Maximum
6. Range (max - min)


````python
marks = np.array([78, 85, 90, 66, 72, 88, 95, 60])
print("mean: ", np.mean(marks))
print("median: ", np.median(marks))
print("variance: ", np.var(marks))
print("standard deviation: ", np.std(marks))
print("minimum: ", np.min(marks))
print("maximum: ", np.max(marks))
print("range: ", np.max(marks) - np.min(marks))
````

# Task 6: Percentiles & Sorting
Using the same marks array:
1. Sort the array
2. Find:
   - 25th percentile
   - 50th percentile
   - 75th percentile
3. Count how many students scored above the average marks

````python
print(marks)
print("sorted marks: ", np.sort(marks))
print("25th percentile: ", np.percentile(marks, 25))
print("50th percentile: ", np.percentile(marks, 50))
print("75th percentile: ", np.percentile(marks, 75))

students_scored_above_average = marks > np.mean(marks)
print("students scored above the average marks: ", students_scored_above_average)
print("number of students scored above the average marks: ", np.sum(students_scored_above_average))
````

# Task 7: Mini Use Case: Sales Analysis
Given daily sales data:
```
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
```
Perform:
1. Total weekly sales
2. Average daily sales
3. Highest and lowest sales day
4. Standard deviation of sales
5. Identify days where sales were above average

````python
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

print("total weekly sales: ", np.sum(sales))
print("average daily sales: ", np.mean(sales))
print("highest sales day: ", np.max(sales))
print("lowest sales day: ", np.min(sales))
print("standard deviation of sales: ", np.std(sales))

sales_above_average = sales > np.mean(sales)
print("days where sales were above average: ", sales_above_average)
````
