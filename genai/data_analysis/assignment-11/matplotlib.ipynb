```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# sales data

```python
sales = pd.read_csv('sales_data.csv')
sales.head()
```

```python
sales.info()
```

```python
sales['Sale_Date'] = pd.to_datetime(sales['Sale_Date'])
sales.head()
```

```python
sales['Month'] = sales['Sale_Date'].dt.month_name()
sales['Year'] = sales['Sale_Date'].dt.year
sales.head()
```

# Task 1: Line Plot (Sales Trend)
1. Create a line plot showing sales trend over months.
2. Add title, X-axis label, Y-axis label.
3. Display the plot.

```python
# average sales per month across both years
monthly = sales.groupby("Month", sort=False)["Sales_Amount"].mean()

plt.figure(figsize=(8, 4))
plt.plot(monthly.index, monthly.values, marker="o")
plt.title("Sales Trend Over Months")
plt.xlabel("Month")
plt.ylabel("Average Sales")
plt.xticks(rotation=45)
plt.show()
```

# Task 5: Stacked Bar Chart
Create a stacked bar chart showing region-wise sales for each month.
Add legend and labels.

```python
stacked = sales.groupby(["Month", "Region"])["Quantity_Sold"].sum().unstack()

months = stacked.index.astype(str)
regions = stacked.columns
bottom = np.zeros(len(months))

plt.figure(figsize=(7, 5))
for region in regions:
    plt.bar(months, stacked[region], bottom=bottom, label=region)
    bottom += stacked[region].values

plt.title("Sales quantity by Region (Stacked by Month)")
plt.xlabel("Month")
plt.ylabel("Sales quantity")
plt.xticks(months, rotation=90)
plt.legend()
plt.show()
```

# Task 7: Pie Chart (Market Share)
1. Create a pie chart representing category share of total sales.
2. Display percentage values.
3. Add title.

```python
share = sales.groupby("Product_Category")["Sales_Amount"].sum()

plt.figure(figsize=(6, 6))
plt.pie(share, labels=share.index, autopct="%1.1f%%")
plt.title("Category Market Share (by Sales)")
plt.show()
```
