### Task 3: Product Pricing (Dictionaries)
- Create a dictionary price_dict where keys are product names and values are prices (integers or floats). Include at least 6 entries.
- Write small code blocks to:

    - Update the price of an existing product.

- Extra (optional): Print the product with both the maximum and minimum prices.

```python
# Show how to get the total number of unique categories using a set.
print(len(categories_set))
```

```python
# Cdictionary price_dict where keys are product names and values are prices 
price_dict = {
    "laptop": 1000,
    "mouse": 20,
    "keyboard": 30,
    "monitor": 150,
    "speaker": 50,
    "headphone": 100
}
print(price_dict)
```

```python
# Add a new product with price to price_dict.
price_dict["speaker"] = 100
print(price_dict)
# Update the price of an existing product.
price_dict["mouse"] = 25
print(price_dict)
# Remove a product by name (handle the case when the product does not exist).
removed = price_dict.pop("keyboard", None)
print("Removed keyboard:" if removed is not None else "keyboard not found:", removed)
print(price_dict)

removed = price_dict.pop("books", None)
print("Removed books:" if removed is not None else "books not found:", removed)
print(price_dict)
```

```python
# Print the average price of all products 
total_price = sum(price_dict.values())
num_products = len(price_dict)
average_price = total_price / num_products
print(f"The average price of all products is: {average_price}")
```

```python
# Print the product with both the maximum and minimum prices.
max_price_product = max(price_dict, key=price_dict.get)
min_price_product = min(price_dict, key=price_dict.get)
print(f"The product with the maximum price is: {max_price_product} with a price of {price_dict[max_price_product]}")
print(f"The product with the minimum price is: {min_price_product} with a price of {price_dict[min_price_product]}")
```

### Task 4: Combined Operations
- Using the products list and price_dict, create a list of tuples named catalog where each tuple is (product_name, price, category).
- From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.
- Print all products that belong to the category that has the maximum number of products.

```python
products = ["laptop", "mouse", "keyboard", "monitor", "speaker", "headphone"]
price_dict = {
    "laptop": 1000,
    "mouse": 25,
    "keyboard": 30,
    "monitor": 150,
    "speaker": 100,
    "headphone": 100,
}
product_categories = {
    "laptop": "electronics",
    "mouse": "electronics",
    "keyboard": "electronics",
    "monitor": "electronics",
    "speaker": "audio",
    "headphone": "audio",
}

print(products)
print(price_dict)
print(categories)
```

```python
catalog = [
    (product, price_dict[product], product_categories[product]) 
    for product in products if product in price_dict]
print(catalog)
```

```python
category_to_products = {}
for product, _, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(product)
print(category_to_products)
```

```python
# Print all products in the category with the maximum number of products
max_category = max(category_to_products, key=lambda cat: len(category_to_products[cat]))
max_category
```
