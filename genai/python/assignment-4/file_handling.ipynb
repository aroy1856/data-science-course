# Task 1: Write Sales Records to a File
- Create a list of sales amounts:
    - sales = [1200, 450, 980, 1500, 3000]
- Open a file named sales_data.txt in write mode.
- Write each sale on a new line in the file.
- Close the file, then reopen it and print its contents.
- Extra (optional): Write the data in comma-separated format instead of separate lines.

````python
sales = [1200, 450, 980, 1500, 3000]

with open('sales_data.txt', 'w') as file:
    for sale in sales:
        file.write(f"{sale}\n")

# reading the file
with open('sales_data.txt', 'r') as file:
    print(file.read())
````

````python
# write the data in comma-separated format instead of separate lines.
with open('sales_data_comma.txt', 'w') as file:
    file.write(','.join(map(str, sales)))
````

# Task 3: Append New Sales
- Append these new sales to the same file:
    - 5000, 2500, 1700
- After appending, reopen and print the entire updated file.
- Extra (optional): Print the total number of lines after appending.

````python
new_sales = [5000, 2500, 1700]
with open('sales_data.txt', 'a') as file:
    for sale in new_sales:
        file.write(f"{sale}\n")

with open('sales_data.txt', 'r') as file:
    print(file.read())
````

# Task 4: Generate Summary Report from File
- Using only file reading operations:
    - Read all sales values from sales_data.txt.
    - Convert them into integers.
    - Calculate and print:
        - Total Sales
        - Highest Sale
        - Lowest Sale
        - Average Sale
- Do not use any advanced libraries.

````python
with open('sales_data.txt', 'r') as file:
    data = file.readlines()
print(data)

sales = [int(s.strip('\n')) for s in data]
print(sales)

# total sales
total_sales = sum(sales)
print(f"Total Sales: {total_sales}")

# highest sale
highest_sale = max(sales)
print(f"Highest Sale: {highest_sale}")

# lowest sale
lowest_sale = min(sales)
print(f"Lowest Sale: {lowest_sale}")

# average sale
average_sale = total_sales / len(sales)
print(f"Average Sale: {average_sale}")
````

# Task 5: Create Product Info File (User Input)
- Ask the user for 3 product names & their prices.
- Write them into a new file products.txt in this format:
    - ProductName | Price
- Read the file and print each line with proper formatting.

````python
products = []
for i in range(3):
    product_name = input("Enter product name: ")
    product_price = input("Enter product price: ")
    products.append(f"{product_name} | {product_price}")

with open('products.txt', 'w') as file:
    for product in products:
        file.write(f"{product}\n")

with open('products.txt', 'r') as file:
    print(file.read())
````


# Task 7: Mini Project - Export Discounted Prices
- Create a dictionary:
```
    prices = {
        "Mouse": 500,
        "Keyboard": 800,
        "Monitor": 7000,
        "Pendrive": 400,
        "Camera": 5000
    }
```
- Ask the user for a discount percentage.
- Write discounted prices into discount_report.txt using:
    - Product | Original Price | Discounted Price
- After writing, read the file and print it to the terminal.
- Extra (optional): Write a summary at the bottom of the file:
    - Total Items: X
    - Average Discounted Price: Y

````python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount_percentage = int(input("Enter discount percentage: "))

with open('discount.txt', 'w') as file:
    for product, price in prices.items():
        discounted_price = price * (1 - discount_percentage / 100)
        file.write(f"{product} | {price} | {discounted_price}\n")

with open('discount.txt', 'r') as file:
    data = file.read()
    print(data)
````

````python
# summary of the file:
total_items = len(data)
discounted_prices = [float(x.split('|')[-1].strip()) for x in data.split('\n') if x]
print(discounted_prices)
average_discounted_price = sum(discounted_prices) / len(discounted_prices)
print(f"Average Discounted Price: {average_discounted_price}")

with open('discount.txt', 'a') as file:
    file.write(f"\nTotal Items: {total_items}\nAverage Discounted Price: {average_discounted_price}")
````
