# Task 1 - Basic Function: Price After Discount
- Write a function apply_discount(price, discount_percent) that:
    - Returns the price after discount.
    - If discount_percent is missing, apply a default discount of 5%.
    - Test the function with:
    - apply_discount(1000, 10)
    - apply_discount(500) # uses default discount    
    - Extra (optional): Add a condition inside the function to ensure discount never exceeds 60%.

```python
def apply_discount(price, discount_percent=5):
    if discount_percent > 60:
        print("Discount cannot exceed 60%")
        return price
    return price * (1 - discount_percent / 100)

print(apply_discount(1000, 10))
print(apply_discount(500))
print(apply_discount(1000, 70)) # should print error message
```

# Task 2 - Recursive Function: Factorial Utility
- Create a recursive function factorial(n) that:
    - Returns the factorial of n.
    - Handles two edge cases: n == 0 and n == 1.
    - Prints an error message if n is negative.
    - Test with:
        - factorial(5)
        - factorial(0)
        - factorial(-3)

```python
def factorial(n):
    if (n < 0):
        print("please enter a positive integer")
        return None        
    elif (n == 0 or n == 1):
        return 1
    return n * factorial(n-1)

print(factorial(5))
print(factorial(0))
print(factorial(-3))
```

# Task 3 - Lambda Function: GST Calculator
- Create a lambda function gst that returns price after adding 18% GST.
- Extra (optional): Create another lambda to compute final price after GST + discount together.


```python
gst = lambda price: price * 1.18
print(gst(100))
```

```python
gst_and_discount = lambda price, discount: (price * (1 - discount / 100)) * 1.18
print(gst_and_discount(100, 10))
```

# Task 4 - Using map(): Apply GST to List of Prices
- Given:
    - prices = [100, 250, 400, 1200, 50]
- Use map() with your gst lambda to generate a new list prices_with_gst.
- Print both lists:
    - Original prices
    - Prices after GST

```python
prices = [100, 250, 400, 1200, 50]
prices_with_gst = list(map(gst, prices))

for i in range(len(prices)):
    print(f"Original price: {prices[i]}, Price after GST: {prices_with_gst[i]}")
```

# Task 5 - Using filter(): Filter Expensive Products
- Given the list:
    - prices = [100, 250, 400, 1200, 50, 2000, 850]
    - Use filter() to:
        - Create a list of prices greater than 500.
        - Create another list of prices less than or equal to 500.
    - Print both lists.

```python
prices = [100, 250, 400, 1200, 50, 2000, 850]
expensive_price = list(filter(lambda x: x > 500, prices))
cheap_price = list(filter(lambda x: x <= 500, prices))
print(expensive_price)
print(cheap_price)
```

# Task 6 - Combined Utility Function
- Write a function process_prices(prices) that:
    - Takes a list of prices.
    - Uses map() + lambda to apply 10% discount to all prices.
    - Uses filter() to keep only discounted prices that are above 300.
    - Returns both lists:
        - discounted_prices
        - filtered_prices
    - Test with:  
        - process_prices([100, 500, 900, 50, 750])

```python
def process_prices(prices):
    discounted_prices = list(map(lambda x: x * 0.9, prices))
    filtered_prices = list(filter(lambda x: x > 300, discounted_prices))
    return discounted_prices, filtered_prices

print(process_prices([100, 500, 900, 50, 750]))
```

# Task 7 - Mini Problem: Menu Using Functions
- Create the following three small functions:
    - add_price(prices_list, price) → adds price to the list.
    - get_average_price(prices_list) → returns average price.
    - get_max_price(prices_list) → returns the maximum price.
    - Then create a simple menu:
        - 1 → Add price
        - 2 → Show average price
        - 3 → Show highest price
        - q → Quit
    - Use only loops + function calls (no OOP).

```python
def add_price(prices_list, price):
    prices_list.append(price)

def get_average_price(prices_list):
    return sum(prices_list) / len(prices_list)

def get_max_price(prices_list):
    return max(prices_list)

prices_list = []

while True:
    print("Menu:")
    print("1 → Add price")
    print("2 → Show average price")
    print("3 → Show highest price")
    print("q → Quit")

    choice = input("Enter your choice: ")
    if choice == 'q':
        break
    elif choice == '1':
        price = input("Enter price: ")
        if price.isdigit():
            add_price(prices_list, int(price))
        else:
            print("Invalid input. Please enter a valid number.")
    elif choice == '2':
        print(f"Average price: {get_average_price(prices_list)}")
    elif choice == '3':
        print(f"Highest price: {get_max_price(prices_list)}")
    else:
        print("Invalid choice. Please try again.")
        continue
```
