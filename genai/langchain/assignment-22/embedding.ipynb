
## PART 2: Document Similarity Search (OpenAI Embeddings)

**Task 4: Similarity Search App (Core Logic)**
1. Convert document chunks into embeddings using OpenAI.
2. Store embeddings in memory or vector store.
3. Create a function:
```
def search(query, top_k=3):
    # returns top_k similar document chunks
```
4. Test similarity search with at least 3 queries.

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
````

````python
def search(query, top_k=3):
    query_embedding = embedding.embed_query(query)
    similarities = cosine_similarity([query_embedding], embedded_docs).flatten()
    top_indices = np.argsort(similarities)[::-1][:top_k]
    return [docs[i] for i in top_indices]
````

````python
queries = [
    "What is the return policy?",
    "What is the shipping policy?",
    "What is the payment policy?",
]

for query in queries:
    best_matches = search(query)
    print(f"Query: {query}")
    print("Best matches:")
    for match in best_matches:
        print(f"- {match.page_content[:100]}...")
    print("-----------------------------------")
````

## PART 4 — Vector Stores

**Task 7: FAISS Vector Store**
1. Store embeddings in FAISS.
2. Perform similarity search.
3. Save FAISS index locally.
4. Reload index and verify search works.

````python
faiss_index = FAISS.from_documents(docs, embedding)

results = faiss_index.similarity_search("What is the return policy?", k=3)
print(results)
````

````python
# save the index to disk
faiss_index.save_local("faiss_index.faiss")

# load the index from disk
faiss_index = FAISS.load_local("faiss_index.faiss", embedding, allow_dangerous_deserialization=True)

results_loaded = faiss_index.similarity_search("What is the return policy?", k=3)
print(results_loaded)
````

````python
print("id before loading: ")
for result in results:
    print(result.id)
    print("-----------------------------------")
print("id after loading: ")
for result in results_loaded:
    print(result.id)
    print("-----------------------------------")
````

**Task 9: FAISS vs ChromaDB Comparison**
Answer briefly:

1. In-memory vs persistent storage → FAISS is mainly an in-memory index (you can `save_local` / `load_local` yourself). Chroma is built as a persistent DB — pass `persist_directory` and it keeps collections on disk automatically.
2. Use cases for FAISS → Fast local similarity search, prototypes, notebooks, high-speed ANN when you just need vectors + scores and are fine managing save/load.
3. Use cases for ChromaDB → Apps that need a real vector DB: persist collections, metadata filters, reload later without rebuilding, multi-session / production-ish local storage.


````python
# Task 9 answers
print("1. FAISS = mostly in-memory (+ manual save/load). Chroma = persistent by design.")
print("2. FAISS: fast local prototypes / ANN search.")
print("3. Chroma: apps that need disk persistence + collections/metadata.")
````


## PART 5 — Mini Project Integration

**Task 10: End-to-End Similarity Search Pipeline**
Build a pipeline:
Documents → Splitter → Embeddings → Vector Store → Similarity Search

Requirements:
- Switchable embedding backend (OpenAI / Hugging Face / Ollama)
- Switchable vector store (FAISS / Chroma)

````python
class DocumentSearchPipeline:
    def __init__(self, embedding_model, vector_store):
        self.embedding_model = embedding_model
        self.vector_store = vector_store

    def embed_documents(self, documents):
        self.vector_store.add_documents(documents)

    def embed_query(self, query):
        return self.embedding_model.embed_query(query)

    def search(self, query, k=3):
        return self.vector_store.similarity_search(query, k=k)

````

````python
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small", dimensions=128)
vector_store = FAISS.from_documents(docs, embedding_model)

document_search_pipeline = DocumentSearchPipeline(embedding_model, vector_store)
````

````python
document_search_pipeline.search("What is the return policy?", k=3)
````

2. Why vector databases are required → Once you have many embeddings, you need fast nearest-neighbor lookup + persistence. A vector DB (or FAISS index) scales search far better than scanning every vector in a Python list.
3. How this pipeline enables RAG systems → Docs → (optional split) → embed → store → similarity_search retrieves the most relevant chunks → those chunks become LLM context. That retrieval step is the “R” in RAG.
