**Task 6: Why Text Splitting is Required (Conceptual)**
Answer briefly:

1. Why large documents cannot be directly passed to LLMs? -> Because LLMs have a context window limitation. even if it can process the whole data the model will hallucinate if we send the whole documents.
2. What problems does chunking solve in GenAI systems? -> It fits text into the context window, improves RAG retrieval (smaller focused chunks rank better), reduces noise/hallucination from unrelated text, and makes embedding + indexing practical.

````python
# Task 6 answers (also in the markdown above)
print("1. Large docs exceed the LLM context window and drown the model in noise → worse answers / hallucinations.")
print("2. Chunking fixes context limits, makes RAG retrieval more precise, and keeps embeddings useful for search.")
````

````python
# Task 8 note — RecursiveCharacterTextSplitter vs CharacterTextSplitter
# Recursive tries separators in order: \n\n → \n → space → "" so it prefers
# keeping paragraphs/sentences together. CharacterTextSplitter only splits on
# one separator (we used "\n"), so it can cut more awkwardly.
print("RecursiveCharacterTextSplitter respects structure better than plain CharacterTextSplitter.")
print("Same chunk_size=100 on the PDF gave similar counts (~394 vs ~395) but recursive")
print("chunks tend to break on paragraph/sentence boundaries when possible.")
````

**Task 10 — Semantic Meaning–Based Splitting (Conceptual + Demo)**

1. Explain what semantic chunking means. 
    -  split where the *topic/meaning* changes, not at a fixed character count. Neighboring sentences that talk about the same idea stay together.
2. Briefly describe how embeddings help semantic splitting.
    - Embeddings turn each sentence into a vector. When cosine similarity between consecutive sentences drops a lot, that drop is treated as a topic boundary → split.
3. (Optional demo) Use any semantic splitter if available.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import re

sample_text = (
    "Machine learning improves from experience using data. "
    "Neural networks are a popular deep learning approach. "
    "Backpropagation updates weights during training. "
    "Meanwhile, e-commerce platforms need shopping carts and databases. "
    "Users add products and checkout with payment gateways. "
    "Order history is stored in relational tables."
)
````

````python
sentences = [s.strip() for s in re.split(r"(?<=[.!?])\s+", sample_text) if s.strip()]
print("Sentences:")
for i, s in enumerate(sentences):
    print(f"  {i}: {s}")

````

````python
vecs = TfidfVectorizer().fit_transform(sentences)
sims = [
    cosine_similarity(vecs[i], vecs[i + 1])[0, 0]
    for i in range(len(sentences) - 1)
]
print("\nAdjacent cosine similarities:", [round(float(x), 3) for x in sims])
````

````python

threshold = float(np.mean(sims))
chunks, current = [], [sentences[0]]
for i, sim in enumerate(sims):
    if sim < threshold:
        chunks.append(" ".join(current))
        current = [sentences[i + 1]]
    else:
        current.append(sentences[i + 1])
chunks.append(" ".join(current))

print(f"\nSemantic-ish chunks (threshold={threshold:.3f}): {len(chunks)}")
for i, c in enumerate(chunks):
    print(f"--- chunk {i+1} ---\n{c}\n")
````

## PART 3 — Mini Integration Task

**Task 11: Unified Preprocessing Pipeline**
Create a function:

```
def load_and_split_documents(path_or_url):
    # loads documents
    # splits them
    # returns chunks
```

Apply it to:

- Local directory
- Web URL


````python
from pathlib import Path
````

````python
def load_and_split_documents(path_or_url, chunk_size=500, chunk_overlap=50):
    """Load from a local path/dir or URL, then split into chunks."""
    path_or_url = str(path_or_url)

    docs = []

    if path_or_url.startswith("http://") or path_or_url.startswith("https://"):
        docs = WebBaseLoader(path_or_url).load()
    else:
        p = Path(path_or_url)
        if p.is_dir():
            docs += DirectoryLoader(p, glob="**/*.txt", loader_cls=TextLoader).load()
            docs += DirectoryLoader(p, glob="**/*.csv", loader_cls=CSVLoader).load()
            docs += DirectoryLoader(p, glob="**/*.pdf", loader_cls=PyPDFLoader).load()
        else:
            raise ValueError(f"Unsupported path: {path_or_url}")

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
    )
    return splitter.split_documents(docs)
````

````python
# Apply to local directory
local_chunks = load_and_split_documents("data")
print(f"Local directory → {len(local_chunks)} chunks")
print(local_chunks[0].page_content[:200])
print("metadata:", local_chunks[0].metadata)
````

````python

# Apply to web URL
web_chunks = load_and_split_documents(
    "https://docs.langchain.com/oss/python/langchain/overview"
)
print(f"Web URL → {len(web_chunks)} chunks")
print(web_chunks[0].page_content[:200])
print("metadata:", web_chunks[0].metadata)
````

**Task 12: Observations & Insights**
Write short answers:

1. Which loader is used for which data type?
2. Best splitter for:
   - Small text
   - Large PDFs
   - Web data
3. Why chunk overlap is important?


1. Loaders by data type:
   - .txt  → TextLoader
   - .csv  → CSVLoader (one Document per row)
   - .pdf  → PyPDFLoader (one Document per page)
   - folder → DirectoryLoader (+ loader_cls / glob per type)
   - webpage → WebBaseLoader
2. Best splitter for:
   - Small text  → CharacterTextSplitter (simple, enough)
   - Large PDFs  → RecursiveCharacterTextSplitter (or page-aware + recursive)
   - Web data    → RecursiveCharacterTextSplitter (messy HTML text needs flexible breaks)
   - Structured md → MarkdownHeaderTextSplitter
   - Meaning-heavy RAG → semantic chunking (embeddings)
3. Why chunk overlap matters:
   Overlap keeps boundary context in both neighboring chunks so retrieval/LLM
   does not lose a sentence that was cut in half at the chunk edge.
