## Task 3: Compare Deployment Platforms

### 1. Streamlit Cloud vs Hugging Face Spaces
| | Streamlit Cloud | Hugging Face Spaces |
|---|---|---|
| Focus | Streamlit apps first | Many SDKs (Streamlit, Gradio, Docker, static) |
| Wiring | Point at GitHub `app.py` + secrets | Space repo / upload + YAML README |
| Secrets | App Settings → Secrets (TOML) | Space secrets / variables |
| Audience | Streamlit / data-app folks | ML / model demo community |

### 2. Pros & cons
**Streamlit Cloud**
- Pros: dead-simple for pure Streamlit; auto-redeploy from GitHub; secrets UX is clear
- Cons: Streamlit-centric; free tier sleeps / resource limits; not ideal for heavy ML runtimes

**Hugging Face Spaces**
- Pros: great for ML demos; Gradio/Docker options; sits next to models/datasets; public discovery
- Cons: cold starts; free hardware is limited; Streamlit setup needs the YAML README dance

### 3. When to use which
- **Streamlit Cloud** — course/data apps that are “just Streamlit + an API key” (like this math agent)
- **HF Spaces** — public ML demos, Gradio UIs, or when you later need Docker / GPU
