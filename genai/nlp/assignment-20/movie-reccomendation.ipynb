````python
# preprocessing genres and keywords
def preprocess_genres(genres):
    if pd.isna(genres) or genres == '[]' or genres == '':
        return ''
    genres_json = json.loads(genres)
    genres_list = [genre['name'] for genre in genres_json]
    return ' '.join(genres_list)

def preprocess_keywords(keywords):
    if pd.isna(keywords) or keywords == '[]' or keywords == '':
        return ''
    keywords_json = json.loads(keywords)
    keywords_list = [keyword['name'] for keyword in keywords_json]
    return ' '.join(keywords_list)

df['genres'] = df['genres'].apply(preprocess_genres)
df['keywords'] = df['keywords'].apply(preprocess_keywords)
df.head()
df.dropna(inplace=True)
````

````python
df.dropna(inplace=True)
df.head()
````

````python
df['movie_blob'] = df['overview'] \
    + ' ' + df['genres'] \
    + ' ' + df['keywords'] \
    + ' ' + df['tagline'] \
    + ' ' + df['title']
data = df[['movie_blob', 'title']]
data.head()
````

**Task 2: Text Preprocessing for Recommendation**
Apply the following steps on the text column:
1. Convert text to lowercase
2. Remove punctuation and special characters
3. Remove stopwords
4. Handle missing values (replace with empty string)

Store cleaned text in a new column: clean_text.

````python
import nltk
````

````python
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
````

````python
stop_words = set(stopwords.words('english'))
wordnet_lemmatizer = WordNetLemmatizer()
````

````python
def nlp_preprocess(text):
    # lowercasing
    text = text.lower()
    # noise removal
    # remove urls
    text = re.sub(r'https?://(www\.)?[-a-zA-Z0-9@:%._\+~#=]{2,256}\.[a-z]{2,4}\b([-a-zA-Z0-9@:%_\+.~#?&//=]*)', '', text)
    # remove email addresses
    text = re.sub(r'[\w\.-]+@[\w\.-]+', '', text)
    # remove html tags
    text = re.sub(r'<.*?>', '', text)
    # remove special characters & emojis
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # remove extra spaces
    text = re.sub(r'\s+', ' ', text)
    # remove numbers
    text = re.sub(r'\d', '', text)
    # stopword removal
    text = ' '.join([word for word in text.split() if word not in stop_words])
    # tokenization
    tokens = word_tokenize(text)
    # lemmatization
    tokens = [wordnet_lemmatizer.lemmatize(word) for word in tokens]
    # join tokens back into text
    return ' '.join(tokens)
````

````python
data['clean_text'] = data['movie_blob'].apply(nlp_preprocess)
data.head()
````

## PART 2 — Text Vectorization

**Task 3: Vectorization using TF-IDF**

1. Use TfidfVectorizer() to convert text into vectors.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
````

````python
vectorizer = TfidfVectorizer(max_features=15000, ngram_range=(1, 3))
````

````python
# Use TfidfVectorizer() to convert text into vectors.
tf_idf_matrix = vectorizer.fit_transform(data['clean_text'])
tf_idf_matrix.shape
````

**Task 4: Similarity Computation**

1. Compute cosine similarity between all items.
2. Store similarity matrix.
3. Explain briefly why cosine similarity is used.

````python
from sklearn.metrics.pairwise import cosine_similarity
import pickle

# Compute cosine similarity between all items
similarity_matrix = cosine_similarity(tf_idf_matrix)

# Store similarity matrix
with open('similarity_matrix.pkl', 'wb') as f:
    pickle.dump(similarity_matrix, f)

# Load similarity matrix
with open('similarity_matrix.pkl', 'rb') as f:
    similarity_matrix = pickle.load(f)
````

## PART 3 — Recommendation Logic

**Task 5: Build Recommendation Function**
Create a function:

```
def recommend(item_name, top_n=5):
    # returns top N similar items
```

Function should:

1. Find index of the selected item
2. Compute similarity scores
3. Sort and return top recommendations

Test with at least 3 different items.

````python
def recommend(item_name, top_n=5):
    # returns top N similar items
    item_index = data[data['title'] == item_name].index[0]
    similarity_scores = list(enumerate(similarity_matrix[item_index]))
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)
    top_n_indices = [i[0] for i in similarity_scores[1:top_n+1]]
    return data['title'].iloc[top_n_indices]
````

````python
print(recommend('The Dark Knight Rises'))
````
