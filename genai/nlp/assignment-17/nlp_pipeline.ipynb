## PART 2 — Advanced Text Cleaning

**Task 3: Removing Noise**
Apply advanced cleaning techniques:

1. Remove URLs
2. Remove email addresses
3. Remove HTML tags
4. Remove special characters & emojis
5. All Other Techniques Mentioned in the Lectures

Store result in: clean_text_advanced.

```python
# url remover
url_pattern = r'https?://(www\.)?[-a-zA-Z0-9@:%._\+~#=]{2,256}\.[a-z]{2,4}\b([-a-zA-Z0-9@:%_\+.~#?&//=]*)'
df['clean_text_advanced'] = df['clean_text_basic'].str.replace(url_pattern, '', regex=True)
# email remover
email_pattern = r'[\w\.-]+@[\w\.-]+'
df['clean_text_advanced'] = df['clean_text_advanced'].str.replace(email_pattern, '', regex=True)
# html tag remover
df['clean_text_advanced'] = df['clean_text_advanced'].str.replace(r'<.*?>', '', regex=True)
# special characters & emojis remover
df['clean_text_advanced'] = df['clean_text_advanced'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)

# remove punctuation
df['clean_text_advanced'] = df['clean_text_advanced'].str.replace(r'[^\w\s]', '', regex=True)


df[['overview', 'clean_text_advanced']].head()
```

**Task 9: Final NLP Pipeline Creation**
Create a single function `nlp_preprocess(text)` that performs:

1. Lowercasing
2. Noise removal
3. Stopword removal
4. Tokenization
5. Lemmatization

Apply this function to the full dataset and store output in final_clean_text.

```python
def nlp_preprocess(text):
    # lowercasing
    text = text.lower()
    # noise removal
    # remove urls
    text = re.sub(r'https?://(www\.)?[-a-zA-Z0-9@:%._\+~#=]{2,256}\.[a-z]{2,4}\b([-a-zA-Z0-9@:%_\+.~#?&//=]*)', '', text)
    # remove email addresses
    text = re.sub(r'[\w\.-]+@[\w\.-]+', '', text)
    # remove html tags
    text = re.sub(r'<.*?>', '', text)
    # remove special characters & emojis
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # remove extra spaces
    text = re.sub(r'\s+', ' ', text)
    # remove numbers
    text = re.sub(r'\d', '', text)
    # stopword removal
    text = ' '.join([word for word in text.split() if word not in stop_words])
    # tokenization
    tokens = word_tokenize(text)
    # lemmatization
    tokens = [wordnet_lemmatizer.lemmatize(word) for word in tokens]
    # join tokens back into text
    return ' '.join(tokens)
```

```python
df['final_clean_text'] = df['overview'].apply(nlp_preprocess)
```

### Task 10 — Observations & Insights

**1. Difference between basic and advanced cleaning**

Basic cleaning is the simple stuff — lowercase, strip punctuation, remove numbers, squeeze extra spaces. That alone makes the text way more consistent, but it doesn't really deal with "noise" that shows up in real web/scraped text.

Advanced cleaning goes further: URLs, emails, HTML tags, special characters/emojis, repeated letters (`soooo` → `so`), slang (`u` → `you`), stopwords. On movie overviews there weren't tons of URLs/emails, but the pipeline still helps and would matter a lot more on tweets/reviews. After advanced cleaning the text is shorter and more about actual content words.

**2. Why lemmatization is preferred over stemming**

Stemming just chops endings with rules (Porter/Snowball). You get weird stems like `studi` / `movi` that aren't real words. Lemmatization (WordNet / spaCy) uses vocab + context and returns actual dictionary forms (`running` → `run`, `better` → `well` with spaCy).

For most NLP tasks lemmatization is nicer because tokens stay readable and you merge variants without mangling them. Stemming is faster/simpler, but the output looks ugly and can merge unrelated words or break related ones inconsistently. I used both — stemming was fine for comparison, but I'd stick with lemmatization in a real pipeline (which is what `nlp_preprocess` does).

**3. Importance of preprocessing in NLP models**

Models don't "understand" messy text the way we do. If you leave casing, punctuation, stopwords, and random noise in, you end up with a huge sparse vocab full of junk (`The` vs `the`, `movie!` vs `movie`). That hurts bag-of-words / TF-IDF style features and wastes capacity in embeddings too.

Preprocessing reduces noise, cuts dimensionality a bit, and makes similar meanings look more alike (`runs`/`running` → `run`). Doesn't magically fix a bad model, but skipping it usually makes everything worse — especially on something like TMDB overviews where length and wording vary a lot.
