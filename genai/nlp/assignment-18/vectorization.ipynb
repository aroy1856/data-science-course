## PART 1 — One-Hot Encoding (Text Level)

**Task 1: Manual One-Hot Encoding**

1. Select 5 short sentences or number you decides for sentences from the dataset.
2. Build a vocabulary of unique words.
3. Create a one-hot encoded matrix manually using Python lists/dictionaries.
4. Display:
   - Vocabulary
   - One-hot vectors for each sentence

```python
dataset =  df['overview'].sample(10).to_list()
```

```python
dataset
```

```python
vocab = set()
for sentence in dataset:
    words = sentence.split()
    for word in words:
        vocab.add(word.lower())
vocab = list(vocab)
len(vocab)
```

```python
word_to_index = {word: index for index, word in enumerate(vocab)}
index_to_word = {index: word for index, word in enumerate(vocab)}
word_to_index
index_to_word
```

```python
len(vocab)
```

```python
one_hot_matrix = []
for sentence in dataset:
    one_hot_sentence = []
    words = sentence.lower().split()
    for word in words:
        one_hot_vector = [0] * len(vocab)
        one_hot_vector[word_to_index[word]] = 1
        one_hot_sentence.append(one_hot_vector)
    one_hot_matrix.append(one_hot_sentence)
one_hot_matrix
```

```python
len(one_hot_matrix[0][0])
```

**Task 2: One-Hot Encoding Using Scikit-learn**

1. Use CountVectorizer(binary=True).
2. Fit and transform the selected sentences.
3. Print:

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer(binary=True)
vectorizer.fit(dataset)
one_hot_matrix = vectorizer.transform(dataset).toarray()
one_hot_matrix
```

```python
vectorizer.get_feature_names_out()
ohe_df = pd.DataFrame(one_hot_matrix, columns=vectorizer.get_feature_names_out())
ohe_df
```

### 8.3 why tfidf downweights common words

so basically tfidf is tf * idf.

tf = how many times the word shows up in that one overview.
idf = if the word is in almost every overview (the, and, of, etc) then idf gets small so the final score goes down.

thats why common words get pushed down even if they appear a lot. rare words that only show in one or two movies keep higher weight cos they actually help tell docs apart.

bow doesnt do this, it just counts. so "the" can dominate everything.

also kinda weird in my notebook — when i summed tfidf across all docs, "the"/"and" were still in the top. i think cos they show up in every row so the sum is still big. the downweighting makes more sense if you look at one document at a time.

**Task 10: Conceptual Questions**
Answer briefly:

1. Difference between One-Hot Encoding and BoW
2. Why N-grams increase dimensionality
3. When to prefer TF-IDF over BoW
4. Limitations of count-based vectorization

### task 10 answers (quick)

**1. one-hot vs bow**

one-hot (binary=True) is just present/not present — 0 or 1. doesnt care how many times.
bow counts how many times each word appears.

same vocab idea, different values. also the manual one-hot in task 1 was per word position which is different from the sklearn doc-level vectors. got a bit confused by that at first.

**2. why ngrams increase dimensionality**

cos now youre storing pairs/triples not just single words. way more unique combos possible so vocab gets huge.
on my 10 samples unigrams < bigrams/trigrams, and (1,2) was bigger than unigrams alone. more features = higher dim.

**3. when to prefer tfidf over bow**

when common words would drown everything and you want distinctive terms. like ranking / basic text classification.
bow is fine for a quick try or if you already removed stopwords and just want raw counts.
for movie overviews id mostly use tfidf.

**4. limitations of count based stuff**

- basically ignores meaning / order (ngrams help a little but not really)
- vectors get fat and sparse
- film and movie treated as totally different words
- no real semantics like embeddings
- typos and rare junk words become their own features which is annoying
